# **PHASE 2- DATA CLEANING**

### REMOVE DUPLICATE ROWS

```python
# --- RULE: Create a copy for cleaning ---
# Never modify the original CSV file.
df_clean = df.copy()

# 1. REMOVE DUPLICATE ROWS
# Check for and remove exact duplicate rows
df_clean = df_clean.drop_duplicates()
print(f"Duplicates removed: {df.shape[0] - df_clean.shape[0]}")
```

### CORRECT DATA TYPES

```python
# 2. CORRECT DATA TYPES
# Convert 'date' column from string/object to datetime format
df_clean['date'] = pd.to_datetime(df_clean['date'])
```

### STANDARDIZE CATEGORICAL LABELS

```python
# 3. STANDARDIZE CATEGORICAL LABELS
# Fix inconsistent labels (e.g., "Male", "male", "  MALE  ") 
# using strip() to remove spaces and title() for uniformity
cat_cols = ['gender', 'region', 'district', 'crop_type', 'education_level', 'experiment_group']
for col in cat_cols:
    df_clean[col] = df_clean[col].str.strip().str.title()
```

### HANDLE MISSING VALUES

```python
# 4. HANDLE MISSING VALUES
# 'farm_size_acres' and 'baseline_yield_kg_per_acre' have missing values.
# Strategy: Filling with the Median (Best for skewed agricultural data)
df_clean['farm_size_acres'] = df_clean['farm_size_acres'].fillna(df_clean['farm_size_acres'].median())
df_clean['baseline_yield_kg_per_acre'] = df_clean['baseline_yield_kg_per_acre'].fillna(df_clean['baseline_yield_kg_per_acre'].median())
```

### OUTLIER IDENTIFICATION (IQR Method)

```python
# 5. OUTLIER IDENTIFICATION (IQR Method)
# Identifying extreme values in 'current_yield_kg_per_acre' and 'net_profit_ghs'
def identify_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return len(outliers)

yield_outliers = identify_outliers(df_clean, 'current_yield_kg_per_acre')
profit_outliers = identify_outliers(df_clean, 'net_profit_ghs')

print(f"Identified {yield_outliers} outliers in Yield and {profit_outliers} in Net Profit.")

# Save the cleaned dataset for the next phase
df_clean.to_csv('agroconnect_cleaned_data.csv', index=False)
```

# **PHASE 4- A/B TESTING (STATISTICAL ANAYLIS**

### Define the Problem and Metric

**Business Problem:** We need to determine if the WHATAPP VOICE NOTE agricultural advisory service
significantly increases farmer yields and net profits compared to SMS Advisory Service.

**Primary Metric:** current_yield_kg_per_acre (Continuous). 

**Secondary Metric:** net_profit_ghs (Continuous).  

### Test 1: Yield Productivity Analysis
This test evaluates the primary impact of the AI-driven advisory on harvest volume.
Metric: current_yield_kg_per_acre

**$H_0$ (Null Hypothesis):** The median current yield is the same for both the control and treatment groups; the Whatapp voice note has no impact on crop productivity.

**$H_1$ (Alternative Hypothesis):** The median current yield for the treatment group is significantly higher than that of the control group. 

**Significance Level ($\alpha$):** $0.05$. 

**Test Justification:** We use the Mann-Whitney U Test because the Shapiro-Wilk test (run in Phase 4) confirmed the yield data is non-normal and skewed

```python
import pandas as pd
import numpy as np
from scipy import stats

# 1. Preparation
df_clean = pd.read_csv('agroconnect_cleaned_data.csv')
df_clean['experiment_group'] = df_clean['experiment_group'].str.strip().str.lower()

yield_control = df_clean[df_clean['experiment_group'] == 'control']['current_yield_kg_per_acre']
yield_treatment = df_clean[df_clean['experiment_group'] == 'treatment']['current_yield_kg_per_acre']
print("Group Summary")
print("=" * 40)

print(f"N yield_control: {len(yield_control)} | N B: {len(yield_treatment)}")
print()

print(f"Mean   : yield_control={yield_control.mean():.2f} | yield_treatment={yield_treatment.mean():.2f}")
print(f"Median : yield_control={yield_control.median():.1f} | yield_treatment={yield_treatment.median():.1f}")
print(f"Std    : yield_control={yield_control.std():.2f} | yield_treatment={yield_treatment.std():.2f}")
print(f"Range  : yield_control=({yield_control.min()}-{yield_control.max()}) | yield_treatment=({yield_treatment.min()}-{yield_treatment.max()})")
print()

print(f"Median lift (yield_treatment−yield_control): {yield_treatment.median() - yield_control.median():.1f} items")
# 2. Assumption Check: Shapiro-Wilk for Normality [cite: 283]
sw_A, p_norm_a = stats.shapiro(yield_control)
sw_B, p_norm_b = stats.shapiro(yield_treatment)

print("Shapiro-Wilk Test")
print("=" * 30)

print(f"A: W={sw_A:.4f}, p={p_norm_a:.4f}")
print(f"B: W={sw_B:.4f}, p={p_norm_b:.4f}")
print()

def check(p, name):
    print(f"{name}: {'NOT normal' if p < 0.05 else 'normal'} (p={p:.4f})")

check(p_norm_a, "yield_control")
check(p_norm_b, "yield_treatment")

print()
print("Decision: Use Mann-Whitney U (non-parametric)")
```

```python
### Step 6b: Mann-Whitney U Test ─────────────────────────────────────────────

u_stat, p_val = stats.mannwhitneyu(yield_control,yield_treatment, alternative='less')

# Effect size (rank-biserial correlation)
r_rb = 1 - (2 * u_stat) / (len(yield_control) * len(yield_treatment))

print("Basket Size — Mann-Whitney U Test")
print("=" * 40)

print(f"A median: {np.median(yield_control):.0f} | B median: {np.median(yield_treatment):.0f}")
print(f"Lift    : {np.median(yield_treatment) - np.median(yield_control):.0f}")
print()

print(f"U: {u_stat:.1f}")
print(f"p: {p_val:.4f}")
print(f"r: {r_rb:.3f}")
print()

decision = "REJECT H0" if p_val < 0.05 else "FAIL TO REJECT H0"
print(f"Decision: {decision}")
```

### Test 2: Financial Profitability Analysis
This test checks if the productivity gains translate into actual financial success after costs.

**Metric:** net_profit_ghs

**$H_0$ (Null Hypothesis):** There is no difference in the net profit generated by farmers in the treatment group compared to the control group.  

**$H_1$ (Alternative Hypothesis):** There is a significant difference in net profit between farmers who received the AI advisory and those who did not.

**Significance Level ($\alpha$):** $0.05$

**Test Justification:** A separate Shapiro-Wilk check was performed on profit data. If $p < 0.05$, the Mann-Whitney U Test is applied to ensure accuracy despite potential outliers or skewness in financial data.

```python
# 1. Preparation
profit_control = df_clean[df_clean['experiment_group'] == 'control']['net_profit_ghs']
profit_treatment = df_clean[df_clean['experiment_group'] == 'treatment']['net_profit_ghs']

print("Group Summary")
print("=" * 40)

print(f"N profit_control: {len(profit_control)} | N B: {len(profit_treatment)}")
print()

print(f"Mean   : profit_control={profit_control.mean():.2f} | profit_treatment={profit_treatment.mean():.2f}")
print(f"Median : profit_control={profit_control.median():.1f} | profit_treatment={profit_treatment.median():.1f}")
print(f"Std    : profit_control={profit_control.std():.2f} | profit_treatment={profit_treatment.std():.2f}")
print(f"Range  : profit_control=({profit_control.min()}-{profit_control.max()}) | profit_treatment=({profit_treatment.min()}-{profit_treatment.max()})")
print()

print(f"Median lift (profit_treatment−profit_control): {profit_treatment.median() - profit_control.median():.1f} items")

# 2. Assumption Check: Shapiro-Wilk for Normality [cite: 283]
sw_a, p_norm_A = stats.shapiro(profit_control)
sw_b, p_norm_B = stats.shapiro(profit_treatment)

print("Shapiro-Wilk Test")
print("=" * 30)

print(f"A: W={sw_a:.4f}, p={p_norm_A:.4f}")
print(f"B: W={sw_b:.4f}, p={p_norm_B:.4f}")
print()

def check(p, name):
    print(f"{name}: {'NOT normal' if p < 0.05 else 'normal'} (p={p:.4f})")

check(p_norm_a, "profit_control")
check(p_norm_b, "profit_treatment")

print()
print("Decision: Use Mann-Whitney U (non-parametric)")
```

```python
# ── Step 6b: Mann-Whitney U Test ─────────────────────────────────────────────

U_stat, P_val = stats.mannwhitneyu(profit_control, profit_treatment, alternative='less')

# Effect size (rank-biserial correlation)
r_rb = 1 - (2 * U_stat) / (len(profit_control) * len(profit_treatment))

print("Basket Size — Mann-Whitney U Test")
print("=" * 40)

print(f"A median: {np.median(profit_control):.0f} | B median: {np.median(profit_treatment):.0f}")
print(f"Lift    : {np.median(profit_treatment) - np.median(profit_control):.0f}")
print()

print(f"U: {U_stat:.1f}")
print(f"p: {P_val:.4f}")
print(f"r: {r_rb:.3f}")
print()

decision = "REJECT H0" if P_val < 0.05 else "FAIL TO REJECT H0"
print(f"Decision: {decision}")
```

### **PHASE 5- BUSINESS RECOMMENDATION**

### Recommendation for Yield Productivity (Primary Metric)

**The Finding:** The analysis returned a p-value of 0.0226. Since this is below the significance threshold of 0.05, the result is statistically significant. This means that we reject the null hypthesis. Hence the Whatsapp voice Note has impact on crop productivity but this impact is less significant (i.e.effect size is 4.2% implying 4.2% on crop productivity).

**Business Implication:** There is a real, measurable difference in yield productivity between the two groups. The WhatsApp Voice advisory service mild increases harvest volume, and this result is not due to random chance.  

**Recommendation:** Scale the Advisory. Because the service is mathematically proven to improve yields, AgroConnect should move forward with the implementation of the voice-based system to replace or supplement SMS.

### Recommendation for Net Profit (Secondary Metric)
**The Finding:** The analysis returned a p-value of 0.7080. This is significantly higher than 0.05, meaning we fail to reject the null hypothesis. This means that there is no difference in the net profit generated by farmers in the treatment group compared to the control group.

**Business Implication:** Although farmers are growing more crops, they are not earning more net profit. This suggests that the cost of following the Whatsapp Voice Note RollOut (e.g., specific inputs or labor) is currently cancelling out the financial gains from the higher yield.

**Recommendation:** Re-calibrate for ROI(Return on Investment). The Whatapp Voice Note Adivsory service content needs to be audited. We must ensure the AI isn't just recommending "better" farming, but "more profitable" farming by suggesting lower-cost, high-impact local inputs.

**Selection Bias:** The test only accounts for farmers with active WhatsApp access, which may not represent the most remote or lower-income segments.
